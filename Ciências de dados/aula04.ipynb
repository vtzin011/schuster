```python
import pandas as pd
import matplotlib.pyplot as plt

# Criando os dados direto no código (substitui o arquivo CSV)
lista_notas = [5.5, 6.0, 7.0, 8.5, 9.0, 4.5, 6.5, 7.5, 8.0, 5.0, 6.8, 7.2, 8.9, 9.5, 3.5, 6.2, 7.8, 8.3, 5.7, 7.1]
dados = pd.DataFrame(lista_notas, columns=['variavel'])

# Histograma
dados['variavel'].hist(color='blue', edgecolor='black')
plt.title('Distribuição da Variável')
plt.xlabel('Valores da Variável')
plt.ylabel('Frequência')
plt.xticks(range(0, 12, 1)) # Ajustado para escala de notas de 0 a 10
plt.grid(True)
plt.show()

# Boxplot
dados.boxplot(column=['variavel'])
plt.show()

# Métricas
media = dados['variavel'].mean()
mediana = dados['variavel'].median()
desvio = dados['variavel'].std()

print("Média:", round(media, 2))
print("Mediana:", round(mediana, 2))
print("Desvio padrão:", round(desvio, 2))

print("\nClassificando a média : ")
if media < 5.0:
    print("Média baixa")
elif media >= 5.0 and media < 7.0:
    print("Média")
elif media >= 7.0:
    print("Média alta \n")

print("\nComparando média e mediana :")
if media > mediana:
    print("Possíveis valores altos extremos")
elif media < mediana:
    print("Possíveis valores baixos extremos")
else:
    print("Distribuição equilibrada")

print("\nClassificando o desvio padrão :")
if media > 7.0 and desvio < 2.0:
    print("Dados altos e consistentes")
elif media > 7.0 and desvio >= 2.0:
    print("Dados altos e dispersos")
else:
    print("Dados não elevados")

print("\nConclusão :")
if desvio < 1.5:
    print("Dados confiáveis")
else:
    print("Dados variáveis")
```
