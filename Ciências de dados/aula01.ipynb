```python
import matplotlib.pyplot as plt

# Dados do gráfico
y = [1, 2, 1, 3, 4, 1, 0]
x = ['segunda', 'terça', 'quarta', 'quinta', 'sexta', 'sabato', 'domingo']

# Criar o gráfico com marcadores e cor azul
plt.plot(x, y, marker="o", color="#0056FF")

# Adicionar títulos e rótulos
plt.title("Horas de estudo")
plt.xlabel("Dias da semana")
plt.ylabel("Horas de estudo")

# Exibir o gráfico
plt.show()
```
